```python
import numpy as np
```

```python
mu = 140.15
R = 700
d = 1500
v = np.sqrt(mu/(d+R))
print(f"Parking velocity: {v:.2f} m/s")
```

```python
ra = 2200
rp = 10
E = 68.08*np.pi/180
a = (ra+rp)/2
e = ra/a - 1
n = np.sqrt(mu/a**3)
t = (E - e*np.sin(E))/n
t_hours = t/3600
print(f"Time since pericenter: {t_hours:.2f} hours")
```

```python
V = np.sqrt(mu*(2/R - 1/a))
print(f"Velocity at r={R} m: {V:.2f} m/s")
```
